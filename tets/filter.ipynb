# Filtrado de Dataset - Producto ID 7850

Este notebook filtra el dataset para mantener únicamente las transacciones del producto con ID **7850**.

```python
import pandas as pd

# Leer el archivo CSV
df = pd.read_csv('dataset_filtrado.csv')

# Filtrar solo las transacciones del producto con ID 7850
df_producto_7850 = df[df['product_id'] == 7850]

# Mostrar información
print(f"Total de registros en el dataset original: {len(df)}")
print(f"Registros del producto 7850: {len(df_producto_7850)}")
print(f"\nInformación del producto:")
if len(df_producto_7850) > 0:
    print(f"  Nombre: {df_producto_7850['product_name'].iloc[0]}")
    print(f"  SKU: {df_producto_7850['product_sku'].iloc[0]}")

# Guardar en un nuevo archivo
df_producto_7850.to_csv('dataset_producto_7850.csv', index=False)
print(f"\n✓ Archivo guardado como 'dataset_producto_7850.csv'")

# Mostrar primeras filas
print("\nPrimeras 5 transacciones:")
df_producto_7850.head()
```
